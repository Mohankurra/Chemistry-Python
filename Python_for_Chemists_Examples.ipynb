```python
import numpy as np
# 'Standard' amount concentration, 1M = 1 mol.dm-3.
c_std = 1
#the concentration of acid.
c = float(input('The concentration of acid is: '))

# The concentration of hydrogen ions at equilibrium(M).
x = np.sqrt(Ka*c*c_std)
x
```

```python
# Temperatures, deC
temperatures = [-20, 0, 25, 40]
for T in temperatures:
  # Amount (mol) and mass (kg) of He in the inflate balloon.
  n_He = p * V / (R * (T + 273.15))
  m_He = M_He/ 1000 * n_He
  # the same amount of air is displaced , but it weights more.
  m_air = M_air/1000 * n_He
  # net force.
  F = (m_air - m_He - m_ballon)*g
  print('{:3d} de C  {:.4f} N'.format(int(T),F))
```

```python
# Gases and thier molar masses
gases = ['He', 'Ne', 'Ar']
gas_masses = [float(input('The molar mass of He is: ')),
             float(input('The molar mass of Ne is: ')),
             float(input('The molar mass of Ar is: '))]
# Temperatures, deC
temperatures = [-20, 0, 25, 40]
for T in temperatures:
  # Amount (mol) and mass (kg) of gas in the inflated balloon.
  n = p * V / (R * (T + 273))
  # The mass of displaced air.
  m_air = n * M_air / 1000
  for gas, M_gas in zip(gases, gas_masses):
    m_gas = n * M_gas / 1000
    # net force.
    F = (m_air - m_gas - m_ballon)*g
    print('{} at  {:3d} de C: force = {:8.5f} N'.format(gas, int(T), F))
```

```python
import numpy as np

# The acid dissociation constant of acetic acid.
Ka = 1.754e-5
# "The standard " amount concentration, 1M = 1 mol.dm-3.
c_std = 1

# The concentration of the acid.
c = float(input('The concentration of the acid is: '))

# Iterate until x changes by less than this amount.
TOL = 1.e-5

x_old  = 0
while True:
  x_new = np.sqrt(Ka * c_std * (c - x_old))
  if abs (x_new - x_old) < TOL:
    break
  x_old = x_new
print(x_new)
print('pH = {:.3f}'.format(-np.log10(x_new)))
```

```python
import numpy as np
def CV_Einstien(T, ThetaE):
  """Return the Einstien molar heat capacity of a material, C_Vm.

  T is the temperature (K) and ThetaE the Einstien temperature for the material being considered. C_Vm is returned in J.K-1.mol-1.

  """

  # Gas constant, J.K-1.mol-1.
  R = 8.314462618

  x =  ThetaE/T
  f = np.exp(x)

  return 3*R*f*(x/(f-1))**2
```

```python
T = 300
ThetaE_Na, ThetaE_dia = 192, 1450
CV_Na, CV_dia = CV_Einstien(T, ThetaE_Na), CV_Einstien(T, ThetaE_dia)
print(f'Material   CVm({T}  K)')
print('{:7s}  {:8.2f}  J.K-1.mol-1'.format('Na', CV_Na))
print('{:7s}  {:8.2f}  J.K-1.mol-1'.format('dia', CV_dia))
```

```python
import numpy as np
# Gas constant, J.K-1.mol-1
R = 8.314462618


def p_ideal(n, T, V):
  ''' ideal gas equation of state , SI units.'''
  return n*R*T/V

def p_Vdw(n, T, V, a, b):
  ''' Vander waals equation of state, SI units.'''
  return n*R*T/(V-n*b) - n**2 * a/V**2

def p_dieterici(n, T, V, a, b):
  ''' Dieterici equation of state, SI units.'''
  return n*R*T*np.exp(-n*a/R/T/V) / (V-n*b)
```

```python
def get_element_symbols(formula):
  '''Return all the element symbols in the string formula.'''

  n = len(formula)
  i = 0
  symbols = []
  # Iterate over the string , nothing the capiital letters, which indicate
  # the start of an element symbol.
  while i<n :
    if formula [i].isupper():
      # A new element symbol
      symbols.append(formula[i])
    elif formula[i].islower():
      # If we encounter a lowercase letter, it is the second character
      # of an element symbol.
      symbols[-1] += formula[i]
    i += 1
  return symbols
```

```python
def get_stiochiometry(formula):
  ''' Return all the elements symbols in the string formula.'''

  n = len(formula)


  # First get all of the "subformulas": each element symbol
  # and its stiochiometry as given in the formuls.
  i = 0
  subformulas = []
  for j in range(1,n):
    # The sentinal for a new element symbol is a capital letter.
    if formula[j].isupper():
      subformula = formula[i:j]
      i = j
      subformulas.append(subformula)
  # The last one needs special treatment because it has no other
  # one following it.
  subformula = formula[i:]
  subformulas.append(subformula)


  # Now build a dictionary of element symbol: stoichiometry pairs.
  stiochiometry = {}
  for s in subformulas:
    # Get the element symbol
    symbol = s[0]
    if len(s)>1:
      if s[1].islower():
        symbol = s[0:2]
      elif s[1].isupper():
        symbol = s[0]
    # What's left after the symbol?
    s_n = s[len(symbol):]
    if not s_n:
      # Nothing? Then the stoichiometry is 1.
      n = 1
    else:
      n = int(s_n)
    # Add to the stoichiomtery  dictionary.
    if symbol in stiochiometry:
      stiochiometry[symbol] += n
    # This is the first time we've encountered this symbol.
    else:
      stiochiometry[symbol] = n
  return stiochiometry
  def get_stiochiometry_formula(formula):
    ''' Prase a chemical formula and return its stoichiometric formula.'''

    stiochiometry = get_stiochiometry(formula)
    # Build the stoichiometry formula string.
    stoichiometry_formula = ''
    for symbol, stoich in stiochiometry.items():
        stoichiometry_formula += symbol
    # Only add the stoichiometry number if it is different from 1.
        if stoich>1:
            stoichiometry_formula += str(stoich)
    return stoichiometry_formula
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.array([-2, -1.5, -1, -0.5, 0.5, 1, 1.5, 2])

y = x**2-x+1
plt.plot(x, y)
plt.show()
#
```

```python
A0, k = 10, 0.5
t = np.linspace(0, 10, 20)
A  = A0*np.exp(-k*t)
plt.plot(t, A, c='r', lw=1, ls=':', marker='o')
plt.xlabel('t/s')
plt.ylabel('[A]/mol.dm^-3')
plt.title('A(t)')
plt.show()
```

```python
A0 = 10
k = [0.2, 0.5, 1]
t = np.linspace(0, 10, 20)
rate_constants = [0.2, 0.5, 1]
for k in rate_constants:
  A = A0*np.exp(-k*t)
  plt.plot(t, A, marker='o',  label=f'k = {k:.1f} s-1'.format(k))
plt.xlabel('t/s')
plt.ylabel('[A]/mol.dm^-3')
plt.title('A(t)')
plt.legend()
plt.show()
# Matplotlib line plots of [A] = [A]0*e^-kt for different rate constants,k.
```

```python
rmax = 10
r = np.linspace(0.1, rmax, 1000)
# Lennard-Jones parameters for Ar: well depth (KJ/mol) and distance at which
# The interaction potential between the two atoms is zero (Angstroms).
E, sigma = 1, 3.4


def LJ(r, E, sigma):
  fac = (sigma/r)**6
  return 4*E*(fac**2-fac)


V = LJ(r, E, sigma)
plt.plot(r, V)
plt.xlabel('r/A')
plt.ylabel('V(r)/ KJ.mol-1')
plt.title('Lennard-Jones Potential for Argon')
plt.show()
#
```

```python
# Start the distance axis at 2 Angstroms.
plt.xlim(2, rmax)
plt.ylim(-1, 1)
plt.plot(r, V)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Boltzmann constant , J.K-1, atomic mass unit (kg)
kB, u = 1.38e-23, 1.66e-27
def fMB (v, T, m):
  """
  Return value of the Maxwell-Boltzmann distribution for a molecule of
  mass m (in kg) at temperature T (in K), at the speed v (in m.s-1).

  """

  fac = m/2/kB/T
  return (fac/np.pi)**1.5*4*np.pi*v**2*np.exp(-fac*v**2)

# A grid of speeds between 0 and vmax.
vmax = int(input('Enter the maximum speed for the Maxwell-Boltzmann distribution (e.g., 2000): '))
v = np.linspace(0, vmax, 1000)
# The molecular mass (in kg) of the N2 molecule.
mN2 = 2*14*u

# Calculate the Maxwell-Boltzmann distribution for a gas of N2 at two different
# temperatures (in K).

T1 = float(input('Enter Temperature 1 (in K, e.g., 300): '))
T2 = float(input('Enter Temperature 2 (in K, e.g., 500): '))
f_N2_T1 = fMB(v, T1, mN2)
f_N2_T2 = fMB(v, T2, mN2)
```

```python
def plot_MB():
  plt.plot(v, f_N2_T1)
  plt.plot(v, f_N2_T2, c='tab:blue')
  plt.xlabel('v (m.s-1)')
  plt.fill_between(v, f_N2_T1, f_N2_T2, color='tab:blue', alpha=0.5, label=f'{T1} K')
  plt.ylabel('f(v)/s.m-1')
  plt.fill_between(v, f_N2_T1, color='tab:blue', alpha=0.2, label=f'{T2} K')
  plt.title('Maxwell-Boltzmann speed Distribution for N2 ')
  plt.legend()
plot_MB()
```

```python
def plot_MB(T, m, label, ls='-'):
  """
  Plot the Maxwell-Boltzmann distribution for an Ideal gas of particles with
  mass m (in kg) at temperature T (in K). Indicate on the plot the mode, mean
  and rms speeds. label is the label to assign to the plotted line and is the linstyle to use.
  """

  # A grid of speeds between 0 and vmax.
  vmax = 1000
  v = np.linspace(0, vmax, 1000)
  # Calculate and plot the Maxwell-Boltzmann distribution.
  f = fMB(v, T, m)
  plt.plot(v, f, label=label, ls=ls, c ='k')

# Calculate the various averages.
  mode, mean, rms = MB_averages(T, m)
# Indicate the averages on the plot.
  fmax = np.max(f)
# Plot straight lines from 0 to the corresponding point on the
# distribution curve for each of the averages.
  plt.plot([mode, mode], [0, fmax], c='red')
  plt.plot([mean, mean], [0, fMB(mean, T,m)], c='blue')
  plt.plot([rms, rms], [0, fMB(rms, T, m)], c='yellow')
  # Add labels: use a bit of trail and error to get the placement right.
  plt.text(mode, fmax*1.05, 'mode', ha='center')
  plt.text(mean, fmax, 'mean')
  plt.text(rms, fmax*0.95, 'rms')
# Set the y-limits from zero to 10% above the distribution maximum.
  plt.ylim(0, 1.1*fmax)
  plt.legend()
  plt.xlabel('T')
  plt.ylabel('fmax')
  plt.title('Maxwell-Boltzmann speed Distribution for Argon and Xenon')
#
```
