```python
# The speed of sound in a gas of molar mass (M) at temperature (T) is c = sqrt(gamma*R*T/M), where R is the gas constant and
#for air the adiabatic index, gamma = 7/5. Estimate the speed of sound in air at
# a) 25 degC, b) -20 degC, take M = 29 gmol-1.


#Solution.

import numpy as np
# The gas constant in J.K-1.mol-1.
R = 8.314

# Mean molar mass of air , in kg.mol-1
M = 29/1000
# The ratio of heat capacities C_p / C_v (adiabatic index) for a diatomic gas
gamma = 7/5  # Set gamma directly as per problem description (7/5 = 1.4)

# Assuming f as a factor to calculate velocity of air at different temperature, in m.s-1.K-1/2.
# c = f(sqrt(T)), f = sqrt(gamma*R/M),
# c = sqrt(gamma*R*T/M)
f = np.sqrt(gamma*R/M)
print(f)
```

```python
# Convert the first temperature from degC to K. a) 25 degC. b) -20 degC
T = float(input('Enter the temperature in degC: '))
T1 = T + 273.1
f*np.sqrt(T1)
```

```python
# Acetic acid (CH3COOH) is a weak acid with pKa = 4.756. What is the pH of a 0.1M solution of acetic acid water?

 # Solution

 # The pKa of acetic acid, log10 of the acid dissociation  constant.
pKa = 4.756         # by definition pKa = -log(Ka).
Ka = 10**-4.756
Ka
```

```python
import numpy as np
# 'Standard' amount concentration, 1M = 1 mol.dm-3.
c_std = 1
#the concentration of acid.
c = float(input('The concentration of acid is: '))

# The concentration of hydrogen ions at equilibrium(M).
x = np.sqrt(Ka*c*c_std)
x
```

```python
pH = -np.log10(x)
pH
```

```python
import numpy as np
# the gas constant in J.K-1.mol-1.
R = 8.314
# The standard molar enthalpy of formation of NH3(g) in kJ.mol-1.
DfH_NH3 = float(input('The standard molar enthalpy of formation of NH3(g) is: '))

# Standard molar entropies for H2(g), N2(g) and NH3(g) in J.K-1.mol-1.
Sm_H2 = float(input('The standard molar entropy of H2(g) is: '))
Sm_N2 = float(input('The standard molar entropy of N2(g) is: '))
Sm_NH3 = float(input('The standard molar entropy of NH3(g) is: '))

# The temperature considered for the equilibrium.
T = float(input('The temperature considered for the equilibrium is: '))
```

```python
# The standard entropies and enthalpies of reaction is.
# If the elements present in either reactants side or products side thier enthalpy of formation can be Zero.
DrH_H2 = 0
DrH_N2 = 0
DrH = 2*DfH_NH3      # Standard enthalpy of reaction in kJ.mol-1.

# Standard entropy of reaction in J.K-1.mol-1
DrS = 2*Sm_NH3 - 3*Sm_H2 - Sm_N2
DrS
DrH,DrS
```

```python
DrG = DrH - T*DrS/1000      # If DrG values are negitive the raection forward to form products.
DrG
```

```python
# DrG standard is equal to -RTlnK. by this if K>>>1 favours products.
K = np.exp(-DrG*1000/(R*T))
K
```

```python
import numpy as np
# The gas constant in J.K-1.mol-1.
R = 8.314
# Molar enthalpy of vaporiszation of water, kJ.mol-1.
DvapH = float(input('The molar enthalpy of vaporiszation of water is: '))
# Normal boiling poiint of water at sea level, k.
T0 = 100 + 275


# Atmospheric scale height, H , in m.
H = float(input('The Atmospheric scale height is: '))

# Height of Everest, m.
z = 8849
# Atmospheric pressure at sea level, atm.
p0 = 1

# Atmospheric pressure at the summit of everest, atm.
p1 = p0*np.exp(-z/H)
p1
```

```python
# The air pressure at the summit, by integrating the claisius-clapeyron equation yields,
# ln(p1/p0)= -DvapH/R(1/T0-1/T1)
# by rearranging above euation, T1 = 1/(1/To-R/DvapH*ln(p1/p0))
# NB convert DvapH from Kj.mol-1 to K.mol-1.
T1 = 1/(1/T0 - R/1000/DvapH*np.log(p1/p0))
T1
```

```python
import numpy as np
# Gas constant, (J.K-1.mol-1) and gravitational acceleration (kg.mol-1).
R = 8.314
g = 9.8
# pressure, Pa
p = float(input('The pressure is: '))
# volume , m3
V = float(input('The volume is: '))
# Mass of uninflated ballon (g). Converting to kg.
m_ballon = float(input('The mass of the uninflated balloon in grams is: ')) / 1000
# Molar mass of helium and mean molar mass of air (g.mol-1).
M_He = float(input('The molar mass of helium is: '))
M_air = float(input('The mean molar mass of air is: '))
```

```python
# Temperatures, deC
temperatures = [-20, 0, 25, 40]
for T in temperatures:
  # Amount (mol) and mass (kg) of He in the inflate balloon.
  n_He = p * V / (R * (T + 273.15))
  m_He = M_He/ 1000 * n_He
  # the same amount of air is displaced , but it weights more.
  m_air = M_air/1000 * n_He
  # net force.
  F = (m_air - m_He - m_ballon)*g
  print('{:3d} de C  {:.4f} N'.format(int(T),F))
```

```python
# Gases and thier molar masses
gases = ['He', 'Ne', 'Ar']
gas_masses = [float(input('The molar mass of He is: ')),
             float(input('The molar mass of Ne is: ')),
             float(input('The molar mass of Ar is: '))]
# Temperatures, deC
temperatures = [-20, 0, 25, 40]
for T in temperatures:
  # Amount (mol) and mass (kg) of gas in the inflated balloon.
  n = p * V / (R * (T + 273))
  # The mass of displaced air.
  m_air = n * M_air / 1000
  for gas, M_gas in zip(gases, gas_masses):
    m_gas = n * M_gas / 1000
    # net force.
    F = (m_air - m_gas - m_ballon)*g
    print('{} at  {:3d} de C: force = {:8.5f} N'.format(gas, int(T), F))
```

```python
print('Net force on ballon(in newtons;  +ve is up)')
print('    T          He            Ne            Ar')
for T in temperatures:
  # Amount (mol) and mass (kg) of gas in the inflated balloon.
  n = p * V / (R * (T + 273))
  # The mass of displaced air.
  m_air = M_air/ 1000*n
  print('{:3d} dec C'.format(int(T)), end='')
  for gas, M_gas in zip(gases, gas_masses):
    m_gas = M_gas/1000*n
    # Net force.
    F = (m_air - m_gas - m_ballon)*g
    print('{:12.5f}'.format(F), end='')
  print()
```

```python
metals = ['Fe', 'Al', 'Hg', 'Ti', 'Pb']
# Standard phase change temperatures for each metal , in K.
melting_point_of_Fe = int(float(input('The melting point of Fe is: ')))
melting_point_of_Al = int(float(input('The melting point of Al is: ')))
melting_point_of_Hg = int(float(input('The melting point of Hg is: ')))
melting_point_of_Ti = int(float(input('The melting point of Ti is: ')))
melting_point_of_Pb = int(float(input('The melting point of Pb is: ')))
```

```python
melting_points = [
    melting_point_of_Fe,
    melting_point_of_Al,
    melting_point_of_Hg,
    melting_point_of_Ti,
    melting_point_of_Pb
]

print('Metal  | Melting Point (K)')
print('----------------------')
for metal, mp in zip(metals, melting_points):
    print(f'{metal:<6s} | {mp:<10d}')
```

```python
boiling_point_of_Fe = int(float(input('The boiling point of Fe is: ')))
boiling_point_of_Al = int(float(input('the boiling point of Al is: ')))
boiling_point_of_Hg = int(float(input('The boiling point of Hg is: ')))
boiling_point_of_Ti = int(float(input('The boiling point of Ti is: ')))
boiling_point_of_Pb = int(float(input('The boiling point of Pb is: ')))
```

```python
boiling_points = [
    boiling_point_of_Fe,
    boiling_point_of_Al,
    boiling_point_of_Hg,
    boiling_point_of_Ti,
    boiling_point_of_Pb

]

print('Metal \t | Boiling Point (K)')
print('----------------------')
for metal, bp in zip(metals, boiling_points):
    print(f'{metal:<6s} | {bp:<10d}')
#
```

```python
# Temperature (K)
T = 500 + 273
for metal, Tm, Tb in zip(metals, melting_points, boiling_points):
  if T > Tb:
    print(f'{metal} is a gas at {T} K')
  elif T > Tm:
    print(f'{metal} is a liquid at {T} K')
  else:
    print(f'{metal} is a solid at {T} K')
```

```python
is_gas = []
for Tb in boiling_points:
  if T>Tb:
    is_gas.append(True)
  else:
    is_gas.append(False)
is_gas
#
```

```python
is_gas = []
for Tb in boiling_points:
  is_gas.append(T>Tb)
is_gas
```

```python
# Even better, we can create this list without having to use the append method.
# list append method
is_gas = [T>Tb for Tb in boiling_points]
is_gas
#
```
